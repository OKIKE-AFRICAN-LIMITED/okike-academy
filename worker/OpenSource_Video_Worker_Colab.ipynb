```python
# 2. Upload worker/render_motion.py (and storyboard-*.json for Offline mode)
from google.colab import files
print('Upload render_motion.py (and your storyboard-*.json file):')
uploaded = files.upload()
```

```python
# 3. Configure Voice & Render Motion-Graphics Video
VOICE = "am_michael"  # am_michael, af_heart, af_bella, bm_george, bf_emma
QUALITY = "m"         # m = 720p30 (~25s), h = 1080p60 (~2m)
SERVER_URL = ""       # Optional: paste your live OpenAcademy URL for auto-upload

import glob, subprocess
if SERVER_URL:
    cmd = ["python", "render_motion.py", "--server-url", SERVER_URL, "--voice", VOICE, "--quality", QUALITY]
else:
    matches = glob.glob("storyboard*.json")
    assert matches, "Upload a storyboard-*.json file downloaded from OpenAcademy!"
    cmd = ["python", "render_motion.py", "--storyboard-json", matches[0], "--output", "lesson_motion.mp4", "--voice", VOICE, "--quality", QUALITY]

subprocess.run(cmd, check=True)
```

```python
# 4. Preview & Download MP4 + VTT Subtitles
import os
from IPython.display import Video, display
from google.colab import files

if os.path.exists("lesson_motion.mp4"):
    display(Video("lesson_motion.mp4", embed=True, width=720))
    files.download("lesson_motion.mp4")
if os.path.exists("lesson_motion.vtt"):
    files.download("lesson_motion.vtt")
```
